```python
historical_df = pd.read_csv("netflix_titles.csv")

current_df = pd.read_csv(
    "netflix_current_top10_movies.csv"
)

print("Historical dataset:", historical_df.shape)
print("Current dataset:", current_df.shape)
```

```python
print("Historical Dataset")
display(historical_df.head())

print("Current Top 10 Dataset")
display(current_df.head())
```

```python
print("Historical Dataset")
display(historical_df.info())

print("Current Top 10 Dataset")
display(current_df.info())
```

```python
historical_df.describe(include="all")
```

```python
current_df.describe(include="all")
```

```python
print(historical_df.isnull().sum())
print("**********")
print(current_df.isnull().sum())
```

```python
historical_df = historical_df.drop_duplicates()

current_df = current_df.drop_duplicates(
    subset=["rank", "title"]
)
```

```python
historical_df.duplicated().sum()
current_df.duplicated().sum()
```

```python
historical_df["release_year"] = pd.to_numeric(
    historical_df["release_year"],
    errors="coerce"
)

current_df["rank"] = pd.to_numeric(
    current_df["rank"],
    errors="coerce"
)

current_df["views"] = pd.to_numeric(
    current_df["views"],
    errors="coerce"
)
```

```python
historical_df["title_clean"] = (
    historical_df["title"]
    .str.strip()
    .str.lower()
)

current_df["title_clean"] = (
    current_df["title"]
    .str.strip()
    .str.lower()
)
```

```python
content_count = historical_df["type"].value_counts()

content_count.plot(
    kind="bar",
    figsize=(7, 5),
    color="firebrick",
    title="Netflix Movies vs TV Shows"
)

plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)
plt.show()
```

```python
release_trend = historical_df[
    "release_year"
].value_counts().sort_index()

plt.figure(figsize=(12, 5))

plt.plot(
    release_trend.index,
    release_trend.values
)

plt.title("Netflix Content Release Trends")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.show()
```

```python
genre_data = historical_df["listed_in"].dropna()

genre_list = genre_data.str.split(", ").explode()

top_genres = genre_list.value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index,
    color="firebrick",
    )

plt.title("Top 10 Genres in Historical Netflix Catalogue")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")
plt.show()
```

```python
country_list = (
    historical_df["country"]
    .dropna()
    .str.split(", ")
    .explode()
)

top_countries = country_list.value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    color="firebrick",
)

plt.title("Top 10 Content-Producing Countries")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

```python
ranked_movies = current_df.sort_values("rank")

plt.figure(figsize=(12, 6))

sns.barplot(
    data=ranked_movies,
    x="views",
    y="title",
    color="firebrick"
)

plt.title("Netflix Current Top 10 Movies by Views")
plt.xlabel("Views")
plt.ylabel("Movie")
plt.show()
```

```python
genre_counts = current_df["genre"].value_counts()

plt.figure(figsize=(9, 5))

genre_counts.plot(
    kind="bar",
    title="Genre Distribution of Current Top 10 Movies",
    color="firebrick"
)

plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45)
plt.show()
```

```python
genre_views = current_df.groupby(
    "genre"
)["views"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=genre_views.values,
    y=genre_views.index,
    color="firebrick"
)

plt.title("Total Views by Movie Genre")
plt.xlabel("Total Views")
plt.ylabel("Genre")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=current_df,
    x="rank",
    y="views",
    hue="genre",
    s=100
)

plt.title("Movie Ranking vs Views")
plt.xlabel("Ranking")
plt.ylabel("Views")
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=current_df,
    x="weeks_in_top10",
    y="title",
    color="firebrick"
)

plt.title("How Long Movies Remained in Top 10")
plt.xlabel("Weeks in Top 10")
plt.ylabel("Movie")
plt.show()
```

```python
historical_movies = historical_df[
    historical_df["type"] == "Movie"
].copy()
```

```python
combined_df = current_df.merge(
    historical_movies[
        [
            "title_clean",
            "release_year",
            "rating",
            "listed_in",
            "country"
        ]
    ],
    on="title_clean",
    how="left"
)

display(combined_df.head(10))
```

```python
unmatched = combined_df[
    combined_df["release_year"].isna()
]

print("Unmatched movies:", len(unmatched))

display(unmatched[["title"]])
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=combined_df.dropna(subset=["release_year"]),
    x="title",
    y="release_year"
)

plt.title("Release Years of Current Top 10 Movies")
plt.xlabel("Movie")
plt.ylabel("Release Year")
plt.xticks(rotation=60)
plt.show()
```

```python
year_views = combined_df.groupby(
    "release_year"
)["views"].sum().sort_index()

year_views.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Current Top 10 Views by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Total Views")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=current_df,
    x="weeks_in_top10",
    y="views"
)

plt.title("Relationship Between Weeks in Top 10 and Views")
plt.xlabel("Weeks in Top 10")
plt.ylabel("Views")
plt.show()
```
